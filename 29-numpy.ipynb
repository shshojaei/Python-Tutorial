# Array Slicing: Accessing Subarrays

1D

```python
# we can use square brackets to access subarrays
# syntax : x[start:stop:step]
# If any of these are unspecified, they default to the values start=0, stop=size of dimension, step=1
x = np.arange(10)
x
```

```python
x[:5] # first five elements
```

```python
x[5:] # elements after index 5
```

```python
x[4:7] # middle subarray
```

```python
x[::2] # every other element
```

```python
x[1::2] # every other element, starting at index 1
```

```python
x[::-1] # all elements, reversed
```

```python
x[5::-2] # reversed every other from index 5
```

2D

```python
x2[:2, :3] # two rows, three columns
```

```python
x2[:3, ::2] # all rows, every other column
```

```python
x2[::-1, ::-1]
```

```python
print(x2[:, 0]) # first column of x2
```

```python
print(x2[0, :]) # first row of x2
```

```python
print(x2[0]) # equivalent to x2[0, :]
```

# Concatenation of arrays

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
np.concatenate([x, y])
```

```python
# You can also concatenate more than two arrays at once:
z = [99, 99, 99]
print(np.concatenate([x, y, z]))
```

```python
 grid = np.array([[1, 2, 3],
                [4, 5, 6]])
# concatenate along the first axis
np.concatenate([grid, grid])
```

```python
# concatenate along the second axis (zero-indexed)
np.concatenate([grid, grid], axis=1)
```

Similarly, np.dstack will stack arrays along the third axis. => must be tuple

```python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                [6, 5, 4]])
# vertically stack the arrays
np.vstack([x, grid])
```

```python
# horizontally stack the arrays
y = np.array([[99],
            [99]])
np.hstack([grid, y])
```

# Splitting of arrays

```python
# pass a list of indices giving the split points
# Notice that N split points lead to N + 1 subarrays
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3, 5])
print(x1, x2, x3)
```

```python
grid = np.arange(16).reshape((4, 4))
grid
```

```python
upper, lower = np.vsplit(grid, [2])
print(upper)
print(lower)
```

```python
left, right = np.hsplit(grid, [2])
print(left)
print(right)
```

# Arithmetic operation (UFunctions)

# Trigonometric functions

```python
theta = np.linspace(0, np.pi, 3)
print("theta = ", theta)
print("sin(theta) = ", np.sin(theta))
print("cos(theta) = ", np.cos(theta))
print("tan(theta) = ", np.tan(theta))
```

```python
x = [-1, 0, 1]
print("x = ", x)
print("arcsin(x) = ", np.arcsin(x))
print("arccos(x) = ", np.arccos(x))
print("arctan(x) = ", np.arctan(x))
```

# Exponents and logarithms

```python
x = [1, 2, 3]
print("x =", x)
print("e^x =", np.exp(x))
print("2^x =", np.exp2(x))
print("3^x =", np.power(3, x))
```

```python
x = [1, 2, 4, 10]
print("x =", x)
print("ln(x) =", np.log(x))
print("log2(x) =", np.log2(x))
print("log10(x) =", np.log10(x))
```

```python
x = [0, 0.001, 0.01, 0.1]
print("exp(x) - 1 =", np.expm1(x))
print("log(1 + x) =", np.log1p(x))
```

```python
# If we’d like to store all the intermediate results of the computation, we can instead use accumulate:
np.add.accumulate(x)
```
