# Combining Datasets: Concat and Append

```python
def make_df(cols, ind):
#Quickly make a DataFrame
    data = {c: [str(c) + str(i) for i in ind]
    for c in cols}
    return pd.DataFrame(data, ind)
# example DataFrame
make_df('ABC', range(3))
```

## Recall: Concatenation of NumPy Arrays

```python
x = [[1, 2],
    [3, 4]]
np.concatenate([x, x], axis=1)
```

```python
ser1 = pd.Series(['A', 'B', 'C'], index=[1, 2, 3])
ser2 = pd.Series(['D', 'E', 'F'], index=[4, 5, 6])
pd.concat([ser1, ser2])
```

```python
df3 = make_df('AB', [0, 1])
df4 = make_df('CD', [0, 1])
print(df3); print(df4); print(pd.concat([df3, df4]))
```

## Duplicate indices

```python
# 1- Catching the repeats as an error
try:
pd.concat([x, y], verify_integrity=True)
except ValueError as e:
print("ValueError:", e)
```

```python
# 2- Ignoring the index
# Sometimes the index itself does not matter, and you would prefer
# it to simply be ignored. You can specify this option using the ignore_index flag. With
# this set to True, the concatenation will create a new integer index for the resulting Series
print(x); print(y); print(pd.concat([x, y], ignore_index=True))  
```

```python
# 3 - Adding MultiIndex keys
print(x); print(y); print(pd.concat([x, y], keys=['x', 'y']))
```

```python
df5 = make_df('ABC', [1, 2])
df6 = make_df('BCD', [3, 4])
print(df5); print(df6); print(pd.concat([df5, df6]))
```

```python
#  By default, the join is a union of the input columns (join='outer'), but we can change this to an intersection of the columns using join='inner':
print(pd.concat([df5, df6], join='inner'))
```

# Combining Datasets: Merge and Join

```python
# example of one-to-one
df1 = pd.DataFrame({'employee': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'group': ['Accounting', 'Engineering', 'Engineering', 'HR']})
df2 = pd.DataFrame({'employee': ['Lisa', 'Bob', 'Jake', 'Sue'],
                    'hire_date': [2004, 2008, 2012, 2014]})
print(df1); print(df2)
```

```python
df3 = pd.merge(df1, df2)
df3
```

```python
# example of many-to-one
df4 = pd.DataFrame({'group': ['Accounting', 'Engineering', 'HR'],
                    'supervisor': ['Carly', 'Guido', 'Steve']})
print(df3); print(df4); print(pd.merge(df3, df4))
```

```python
# example of many-to-many
df5 = pd.DataFrame({'group': ['Accounting', 'Accounting','Engineering', 'Engineering', 'HR', 'HR'], 
                    'skills': ['math', 'spreadsheets', 'coding', 'linux','spreadsheets', 'organization']})
print(df1); print(df5); print(pd.merge(df1, df5))
```

```python
 print(df1); print(df2); print(pd.merge(df1, df2, on='employee'))
```

```python
df3 = pd.DataFrame({'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'salary': [70000, 80000, 120000, 90000]})
print(df1); print(df3);
print(pd.merge(df1, df3, left_on="employee", right_on="name"))
```

```python
pd.merge(df1, df3, left_on="employee", right_on="name").drop('name', axis=1)
```

```python
df1a = df1.set_index('employee')
df2a = df2.set_index('employee')
print(df1a); print(df2a)
```

```python
# You can use the index as the key for merging by specifying the left_index and/or
# right_index flags in pd.merge():

print(df1a); print(df2a);
print(pd.merge(df1a, df2a, left_index=True, right_index=True))
```

```python
print(df1a); print(df2a); print(df1a.join(df2a)) #  DataFrames implement the join() method, which performs a merge that defaults to joining on indices:
```

## Specifying Set Arithmetic for Joins

```python
df6 = pd.DataFrame({'name': ['Peter', 'Paul', 'Mary'],
                    'food': ['fish', 'beans', 'bread']},
                    columns=['name', 'food'])
df7 = pd.DataFrame({'name': ['Mary', 'Joseph'],
                    'drink': ['wine', 'beer']},
                    columns=['name', 'drink'])
print(df6); print(df7); print(pd.merge(df6, df7))
```

```python
pd.merge(df6, df7, how='inner') # intersection of the two sets of inputs
```

```python
print(pd.merge(df6, df7, how='outer')) # returns a join over the union of the input columns, and fills in all missing values with NAs
```

```python
print(pd.merge(df6, df7, how='left')) 
```

```python
print(pd.merge(df6, df7, how='right'))
```

## Overlapping Column Names: The suffixes Keyword

```python
df8 = pd.DataFrame({'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'rank': [1, 2, 3, 4]})
df9 = pd.DataFrame({'name': ['Bob', 'Jake', 'Lisa', 'Sue'],
                    'rank': [3, 1, 4, 2]})
print(df8); print(df9); print(pd.merge(df8, df9, on="name"))
```

```python
print(pd.merge(df8, df9, on="name", suffixes=["_L", "_R"]))
```

## Aggregate, filter, transform, apply

```python
# GroupBy objects have aggregate(), filter(), transform(), and apply() methods
rng = np.random.RandomState(0)
df = pd.DataFrame({'key': ['A', 'B', 'C', 'A', 'B', 'C'],
                   'data1': range(6),
                   'data2': rng.randint(0, 10, 6)},
                    columns = ['key', 'data1', 'data2'])
df
```

### Aggregation

```python
df.groupby('key').aggregate(["min", "median" , "max"])
```

```python
 df.groupby('key').aggregate({'data1': 'min',
                              'data2': 'max'})
```

### Filtering

```python
def filter_func(x):
    return x['data2'].std() > 4
    
print(df);
print(df.groupby('key').std());
print(df.groupby('key').filter(filter_func))
```

### Transformation

```python
df.groupby('key').transform(lambda x: x - x.mean())
```

### Apply

```python
def norm_by_data2(x):
    # x is a DataFrame of group values
    x['data1'] /= x['data2'].sum()
    return x

print(df.groupby('key').apply(norm_by_data2))
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
titanic = sns.load_dataset('titanic')
```

```python
titanic.head()
```

```python
titanic.groupby('sex')[['survived']].mean()
```

```python
titanic.groupby(['sex', 'class'])['survived'].aggregate('mean').unstack()
```

## Pivot Table Syntax

```python
titanic.pivot_table('survived', index='sex', columns='class')
```

## Multilevel pivot tables

```python
age = pd.cut(titanic['age'], [0, 18, 80])
titanic.pivot_table('survived', ['sex', age], 'class')
```

```python
fare = pd.qcut(titanic['fare'], 2)
titanic.pivot_table('survived', ['sex', age], [fare, 'class'])
```

```python
titanic.pivot_table(index='sex', columns='class', aggfunc={'survived':sum, 'fare':'mean'})
```
