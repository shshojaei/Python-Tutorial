- Python has a set of built-in math functions, including an extensive math module, that allows you to perform mathematical tasks on numbers.

# Built-in Math Functions

```python
x = min(5, 10, 25)
print(x)
```

```python
y = max(5, 10, 25)
print(y)
```

```python
x = abs(-7.25)
print(x)
```

```python
x = pow(4, 3)
print(x)
```

# The Math Module

Python has also a built-in module called math, which extends the list of mathematical functions.

```python
import math

x = math.sqrt(64)
print(x)
```

```python
import math

x = math.ceil(1.4)
print(x) 
```

```python
import math

y = math.floor(1.4)
print(y) 
```

```python
import math

x = math.pi
print(x)
```
